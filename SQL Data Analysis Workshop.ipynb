## Level 2: Intermediate – Aggregations and Basic Joins


### Task 4: Join Orders with Customers


#### Task 4.1: Retrieve customer names along with their order IDs.

```python
cursor.fetchall()
cursor.execute(
"""
SELECT customers.customer_id , first_name 
from customers 
LEFT JOIN orders ON customers.customer_id = orders.customer_id

"""
)
res= cursor.fetchall()
res
```

#### Task 4.2: Retrieve Customer Names with Total Orders

```python
cursor.execute(
"""
SELECT customers.first_name , orders.order_id
FROM customers
LEFT JOIN orders ON customers.customer_id = orders.customer_id
GROUP BY (customers.first_name )
"""
)

res= cursor.fetchall()
res
```

### Task 5: Retrieve products with stock quantities less than 50.

```python
cursor.execute(
"""
SELECT products.product_name , stocks.stock_quantity 
FROM products
LEFT JOIN stocks ON products.product_id = stocks.product_id
WHERE stocks.stock_quantity <50
"""
)
res= cursor.fetchall()
res
```

## Level 3: Advanced – Complex Joins and Subqueries
(using Pandas read_sql)

### Task 1: Retrieve the Top 5 Products with the Highest Total Sales Revenue


```python
query ="""

SELECT P.product_name , order_items.list_price, quantity,
SUM( order_items.list_price * order_items.quantity ) AS Total_rev
FROM products AS P 
LEFT JOIN order_items ON P.product_id = order_items.product_id
GROUP BY P.product_name
ORDER BY Total_rev DESC 
LIMIT 5;

"""

df=pd.read_sql(query,connect)
df.dropna(inplace=True)
df
```

### Task 2: Find the Customers Who Placed the Most Orders in June, November 2018

### Task 3: List All Stores with Their Total Stock Quantities for All Products

```python
import seaborn as sns 
query="""
SELECT 
    stores.store_name ,
    products.product_name ,
    stocks.quantity 
FROM stores 
LEFT JOIN stocks 
    ON stores.store_id = stocks.store_id 
LEFT JOIN products
    ON stocks.product_id = products.product_id
GROUP BY stores.store_name
"""

df=pd.read_sql(query,connect)
sns.barplot(data=df, x='store_name', y='quantity', estimator=sum)
```

### Task 1: Find the Store with the Highest Total Revenue
Retrieve the store name and total revenue (sum of list_price * quantity) across all orders.




```python
query ="""
SELECT stores.store_name,order_items.list_price, quantity,
SUM( order_items.list_price * order_items.quantity ) AS Total_rev
FROM orders  
LEFT JOIN order_items ON orders.order_id = order_items.order_id
LEFT JOIN stores ON  stores.store_id = orders.store_id 
GROUP BY stores.store_name
ORDER BY Total_rev DESC 
LIMIT 1;

"""

df=pd.read_sql(query,connect)
df.dropna(inplace=True)
df
```
